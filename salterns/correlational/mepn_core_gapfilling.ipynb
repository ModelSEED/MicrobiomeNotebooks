```python
# Load ATP correction gapfill data for each MePn MAG
from tqdm import tqdm
import json
import os

atp_data = {"MePn_consumers": {}, "BET_consumers": {}}
atp_dir = '../cliff_mags/mac/salterns/atp'
for name, mems in zip(list(atp_data.keys()), [MePn_consumers, BET_consumers]):
    for i in tqdm(sorted(mems)):
        atp_file = os.path.join(atp_dir, f'{i}__.RAST.json')
        if os.path.exists(atp_file):
            with open(atp_file, 'r') as fh:
                atp_data[name][i] = json.load(fh)
    else:
        print(f'ATP data not found: {i}')

print(f'Loaded ATP data for {len(atp_data["MePn_consumers"])} / {len(MePn_consumers)} MAGs')
print(f'Loaded ATP data for {len(atp_data["BET_consumers"])} / {len(BET_consumers)} MAGs')
```

```python
# Count core gapfilled reactions per condition (excluding exchange reactions)
genome_core_gf = {"MePn_consumers": {}, "BET_consumers": {}}
for name, mems in zip(list(atp_data.keys()), [MePn_consumers, BET_consumers]):
    for genome_id in atp_data[name]:
        genome_core_gf[name][genome_id] = {}
        _data = atp_data[name][genome_id]
        for cond_id in _data['gapfill']:
            _gapfill_data = _data['gapfill'][cond_id]
            if _gapfill_data['target']:
                _gap_fill_add = list(filter(lambda x: x[:3] != 'EX_', _gapfill_data['new']))
                genome_core_gf[name][genome_id][cond_id] = len(_gap_fill_add)

print(f'Processed {len(genome_core_gf["MePn_consumers"])} MePn-consuming genomes')
print(f'Processed {len(genome_core_gf["BET_consumers"])} BET-consuming genomes')
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import matplotlib.patches as patches
import numpy as np
from scipy.cluster import hierarchy
from scipy.spatial.distance import pdist

for name in atp_data.keys():
    d = genome_core_gf[name]

    df = pd.DataFrame.from_dict(d, orient='index')
    df = df.drop(columns=[col for col in df.columns if ("." in col and "O2" not in col)])
    df_clean = df.fillna(0)

    stats = {metID: {"max": row.max(), "median": row.median(), "min": row.min(), "total": row.sum()}
            for metID, row in df.T.iterrows()}
    stats_df = pd.DataFrame.from_dict(stats, orient='index')
    display(stats_df.sort_values(by="total"))

    # Precompute clustering order and reindex
    row_order = hierarchy.leaves_list(hierarchy.linkage(pdist(df_clean.values), method='average'))
    col_order = hierarchy.leaves_list(hierarchy.linkage(pdist(df_clean.values.T), method='average'))
    df_clustered = df_clean.iloc[row_order, :].iloc[:, col_order]

    g = sns.clustermap(
        df_clustered,
        annot=True,
        cmap='viridis',
        annot_kws={"size": 20},
        figsize=(24, 16),
        row_cluster=False,      # already ordered
        col_cluster=False,      # already ordered
        xticklabels=True,
        yticklabels=True,
    )

    ax = g.ax_heatmap
    ax.set_title(f"Core Gap Filling - {name}", fontsize=20, pad=12)
    ax.set_ylabel("MAG", fontsize=14)
    ax.set_xlabel("Conditions", fontsize=14)
    ax.set_xticklabels(ax.get_xticklabels(), fontsize=14, rotation=70, ha='right')
    ax.set_yticklabels(ax.get_yticklabels(), fontsize=10, rotation=0)

    if 'Llac' in df_clustered.columns:
        dlac_idx = list(df_clustered.columns).index('Llac')
        rect = patches.Rectangle((dlac_idx, 0), 1, len(df_clustered), linewidth=3, edgecolor='red', facecolor='none')
        ax.add_patch(rect)

    g.figure.savefig(f'core_gap_filling_{name}.png', dpi=800, bbox_inches='tight')
    plt.show()
    print(f'Saved core_gap_filling_{name}.png')
```
